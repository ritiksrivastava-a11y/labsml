## **Q1.5 — Score the Model (MAE, MSE, RMSE, R²)**

**Idea:** Instead of looking at every residual, we use four numbers to judge the model.

| Name | What it means in simple words |
|---|---|
| **MAE** | On average, how many lakh the guess is off by |
| **MSE** | The squared errors averaged (big mistakes count extra) |
| **RMSE** | Square root of MSE, so it is in lakh again. Easier to read than MSE |
| **R²** | A score up to 1. Closer to 1 means the model explains the prices well |
